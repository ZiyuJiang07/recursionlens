# RecursionLens in JupyterLab

One recursive function, three views: an invocation tree, a compressed subproblem DAG, and a naive-versus-memoized comparison. No Graphviz or network access is required.

```python
from recursionlens import trace

@trace
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)
```

## 1. The actual call tree

Each node is one observed invocation. Coral badges mark argument signatures seen before. Expand or collapse nodes to inspect the tree.

```python
naive = fib.run(5)
naive
```

## 2. Show every call at once

Expanded mode lays out all 15 invocations as one SVG tree. Nothing is collapsed. Repeated calls are underlined coral.

```python
naive.show(mode="expanded")
```

## 3. Compress identical subproblems into a DAG

The tree has 15 nodes, but only six unique argument signatures. The DAG exposes that difference directly.

```python
naive.show(mode="dag")
```

## 4. Measure what memoization changes

The second run uses the same function and input. The comparison checks the returned values and reports executed bodies separately from cache hits.

```python
memoized = fib.with_options(memoize=True).run(5)
comparison = naive.compare(memoized)
comparison
```

## 5. Export standalone evidence

These files are written beside the notebook and can be opened without JupyterLab.

```python
naive.export_html("fib-tree.html")
naive.export_html("fib-dag.html", mode="dag")
naive.export_html("fib-expanded.html", mode="expanded")
naive.export_json("fib-trace.json")
comparison.export_html("fib-comparison.html")
```

## Scope

Version 0.1 targets synchronous, deterministic teaching functions. Memoization requires hashable arguments. RecursionLens executes the decorated code and is not a sandbox for untrusted programs.
